## 1. Comparing each function with n²

For f(n)=O(n²), there must be constants c>0 and n₀ such that f(n)≤cn² for every integer n≥n₀. For f(n)=Ω(n²), the inequality is reversed: cn²≤f(n). The function is Θ(n²) exactly when both conditions hold.

- **5n²+3n:** O(n²), since for every n≥1, 5n²+3n≤8n² (c=8,n₀=1). Ω(n²), since 5n²≤5n²+3n (c=5,n₀=1). Therefore Θ(n²).
- **n²/1000:** O(n²) and Ω(n²); both inequalities hold with c=1/1000 and n₀=1. Therefore Θ(n²).
- **n log₂ n:** O(n²), since log₂ n≤n for every integer n≥1 (c=1,n₀=1). It is not Ω(n²): for any proposed c>0,n₀, choose n≥n₀ large enough that log₂(n)/n<c. Then n log₂ n<cn², contradicting the required lower inequality. Thus it is not Θ(n²).
- **n² log₂ n:** Ω(n²), since log₂ n≥1 for every integer n≥2 (c=1,n₀=2). It is not O(n²): for any proposed c>0,n₀, choose n≥n₀ with log₂ n>c. Then n² log₂ n>cn², contradicting the required upper inequality. Thus it is not Θ(n²).
- **4 log₂ n:** O(n²), since log₂ n≤n² for every integer n≥1 (c=4,n₀=1). It is not Ω(n²): for any proposed c>0,n₀, choose n≥n₀ large enough that 4 log₂(n)/n²<c. Then 4 log₂ n<cn², a contradiction. Thus it is not Θ(n²).

Changing the base of the logarithm only changes it by a constant factor and does not change these classifications.

## 3. Binary search cases

Assume a sorted array of n distinct elements and count comparisons with array elements.

- **Best: Θ(1).** The target is the middle element checked first, requiring one comparison.
- **Worst: Θ(log n).** The target is absent, or is at a position reached only after repeatedly following a deepest branch. Each comparison halves the remaining interval.
- **Average: Θ(log n).** Choose the target uniformly at random from the n array elements. Its expected search depth in the balanced decision tree is Θ(log n).

## 4. Transitivity of O

Suppose f=O(g) and g=O(h). By the first assumption, there are c₁>0,n₁ such that |f(n)|≤c₁|g(n)| for every n≥n₁. By the second, there are c₂>0,n₂ such that |g(n)|≤c₂|h(n)| for every n≥n₂. Let n₀=max(n₁,n₂). For every n≥n₀, |f(n)|≤c₁|g(n)|≤c₁c₂|h(n)|. Since c₁c₂>0, c=c₁c₂ and this n₀ satisfy the definition of f=O(h). Thus O is transitive.

## 5. Why Θ is the statement we want and O is the one we usually get

Θ describes an asymptotically tight rate of growth: it bounds a function both above and below by constant multiples of the same comparison function. Saying an algorithm takes Θ(n log n) time means that, for sufficiently large inputs, its running time grows on that order up to constant factors. This is more informative than an upper bound alone because it rules out substantially smaller as well as larger growth rates. In practice, however, analysis often first gives an O bound. An upper bound can be obtained by counting a safe maximum number of operations, while a matching lower bound requires proving that enough work is unavoidable. That lower-bound argument may require additional reasoning about the algorithm or the problem, so it is common to establish only that an algorithm takes at most O(g(n)) time. Big-O is also useful when a tight bound is unknown or unnecessary. The distinction matters: an O(g(n)) statement does not automatically imply Θ(g(n)). We want Θ when we know the tight asymptotic behavior; O is the more readily established guarantee that running time will not grow faster than the stated rate, apart from constant factors.
